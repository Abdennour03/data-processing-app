### Step 1: Standardize the Data

which contains  
$n = 200$ observations and $d = 16$ features.

The data matrix is:

$$
X  \in \mathbb{R}^{200 \times 16}
$$


#### Feature-wise Mean and Standard Deviation

For each feature $j \in \{1, \dots, 16\}$, we compute the mean:

$$
\mu_j = \frac{1}{200} \sum_{i=1}^{200} x_{ij}
$$

```python
mu_j = np.mean(X, axis=0)
```

and the standard deviation:

$$
\sigma_j = \sqrt{\frac{1}{200} \sum_{i=1}^{200} (x_{ij} - \mu_j)^2}
$$

```python
sigma_j = np.std(X, axis=0)
```

#### Standardization (Z-score Normalization)

Each value of the dataset is standardized using:

$$
z_{ij} = \frac{x_{ij} - \mu_j}{\sigma_j}
$$

```python
Z = (X - mu_j)/sigma_j
print("Z-score = ", Z)
```

#### Properties After Standardization

After this transformation, each feature satisfies:

$$
\frac{1}{200} \sum_{i=1}^{200} z_{ij} = 0
\quad \text{and} \quad
\frac{1}{200} \sum_{i=1}^{200} z_{ij}^2 = 1
$$

This means that all features have zero mean and unit variance.

#### Goal of PCA

Because the data lies in a high-dimensional space ($16$ dimensions), it cannot be visualized directly.  
The goal of Principal Component Analysis (PCA) is to project the data into a lower-dimensional space:


### Step 2: Covariance Matrix Computation



Let the standardized data matrix be:

$$
Z \in \mathbb{R}^{200 \times 16}
$$

The covariance matrix \(C\) is defined as:

$$
C = \frac{1}{n-1} Z^T Z
$$


Each element \(C_{ij}\) represents the covariance between feature \(i\) and feature \(j\):

$$
C_{ij} = \frac{1}{n-1} \sum_{k=1}^{n} z_{ki} z_{kj}
$$




```python
n = Z.shape[0]
print('n =', n)
```

```python
C = (Z.T @ Z) / (n - 1)
print("Cov = ", C)
```

### Step 3: Eigenvalues and Eigenvectors

```python
eigenvalues, eigenvectors = np.linalg.eig(C)
```

### Step 4: Sort Eigenvalues and Eigenvectors

```python
idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[idx]

print("Landa_i =", eigenvalues)
print("-----")
print('V_i =', eigenvectors)
```

### Step 5: Choose Number of Components

```python
explained_variance_ratio = eigenvalues / np.sum(eigenvalues)
cumulative_variance = np.cumsum(explained_variance_ratio)

cumulative_variance
```

### Step 6: Projection onto Principal Components

```python
k = 2  # for visualization or dimensionality reduction

W_k = eigenvectors[:, :k]
Y = Z @ W_k
```
